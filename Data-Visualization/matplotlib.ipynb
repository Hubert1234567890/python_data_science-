```python
import matplotlib.pyplot as plt
x =[1,2,3,4,5]
y =[1,4,9,16,25]
plt.plot(x,y)
```

```python
x =[1,2,3,4,5]
y =[1,4,9,16,25]
plt.plot(x,y)
plt.xlabel("x axis ")
plt.ylabel("y axis ")
plt.title("basic line plot ")
```

```python
x =[1,2,3,4,5]
y =[1,4,9,16,25]
plt.plot(x,y ,color = 'red',linestyle ='--',marker='o',linewidth = 3,markersize = 10)
plt.grid(True)
```
