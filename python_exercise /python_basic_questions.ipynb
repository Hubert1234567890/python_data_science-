```python
def fact(n):
    if n == 0:
        return 1
    else :
        return n * fact(n-1)

num = int(input("enter the number: "))
print(f"factorial of the number {num} is {fact(num)}")
    
```

**Question 16:** Write a Python program to find the sum of the first n natural numbers.

```python
def sum(n):
    if n==0:
        return 0
    else :
        return n + sum(n-1)

num = int(input("enter the numebr: "))
print(f"the sum of first {num} natural numbers is {sum(num)} ")
```

**Question 17:** Write a Python program to check if a year is a leap year.

```python
def leap_year_check(n):
    if n % 4 == 0:
        return 'leap_year'
    else :
        return 'not a leap_year'

year = int(input("enter the year you want to check: "))
print(f"the given year {year} is {leap_year_check(year)}")
```

**Question 19:** Write a Python program to check if a string is a palindrome.

```python
def palindrom_check(string):
    if string == string[::-1]:
        return 'palindrome'
    else: 
        return 'not a palindrome'

string = input("enter any string: ")
print(f"the given string {string} is {palindrom_check(string)}")
```
