14  Write a function that takes two lists and returns a new list containing only the elements that are present in both lists. Print the intersected list.

```python
def intersection_list(lst1,lst2):
    intersection =[]
    for i in range(len(lst1)):
        for j in range(len(lst2)):
            if lst1[i]==lst2[j]:
                intersection.append(lst1[i])
    return intersection            
          
num1 = input("enter first list : ").split()
num2 = input("enter second list : ").split() 
print(num1)
print(num2)
print(f"the requried intersection list is {intersection_list(num1,num2)}")
```

15 Write a function that rotates a list by n positions. Print the original and rotated lists.

```python
def rotated_list(lst,n):
    length  = len(lst)
    first_half = lst[0:length-n]
    rotated_half = lst[length-n:]
    new_list = rotated_half + first_half 
    return new_list

num = input("enter the list with spaces:").split()

print(f"{rotated_list(num,2)}")
```
