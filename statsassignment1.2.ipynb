lets now carray out the whole operation for without replacememt case 

```python
import itertools
import numpy as np 
import matplotlib.pyplot as plt 

## using the population 
population  = np.array([6,8,10,12,14])

pop_mean = np.mean(population)
pop_var = np.var(population)

## darwing samples with replacemet 
samples =[]

for sample in itertools.combinations(population , 3):
    samples.append(sample)

sample = np.array(samples)
 #this is the array containing all the samples of sizes 3 
```

lets now calculate the sample statistic for each sample namely sample mean and smaple varaince 

```python
# calculating sample mean 

sample_mean=[]

for i in samples:
    mean = np.mean(i)
    sample_mean.append(mean)

x_bar = np.array(sample_mean)  ## this array conatins  the means each samples  

# calcluating sample variance 

sample_variance=[]

for i in samples:
    variance = np.var(i,ddof=1)
    sample_variance.append(variance)

s_square = np.array(sample_variance)  ## this array contains the sample variances of each samples 


```

lets now do some calculation  on the x_bar and s_square array 

```python
mean_of_x_bar = np.mean(x_bar)
variance_of_x_bar = np.var(x_bar)

standard_error_of_x_bar = np.sqrt(variance_of_x_bar)

print(f"the expected value of the x_bar is {mean_of_x_bar}")
print(f"the true population mean is {pop_mean}")
print(f"variance of x_bar is {variance_of_x_bar}")
print(f"the standard error of x_bar is {standard_error_of_x_bar}")
```

since the expected value of x bar is equal to the true population mean the result is consistent with the theory 

```python
mean_of_s_sqaure = np.mean(s_square)
varianec_of_s_square = np.var(s_square)

standard_error_of_s_square = np.sqrt(varianec_of_s_square)

print(f"the expected value of the s_square is {mean_of_s_sqaure}")
print(f"the true population variance  is {pop_var}")
print(f"variance of s_square is {varianec_of_s_square}")
print(f"the standard error of s_square is {standard_error_of_s_square}")
```

we can see that in the case of (without replacement) the expected value of s_square is not equal to the population varaince 
with is true  in (with replacement) case
