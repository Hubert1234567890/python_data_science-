so the porblem is that i have been give a population {6,8,10,12,14} and i m supposed to draw all the possible samples of sizes 3 
with replacemet and without replacement  then  calculate the sample mean and variance of each samples respectievely and then calculate their 
mean , variance and standard error and comapre it with the population parameters . 

first we will do the with replacemet part . For that we will be importing a module itertools 
also we will be using the libaries like numpy and matplotlib 

```python
import itertools
import numpy as np 
import matplotlib.pyplot as plt 

## using the population 
population  = np.array([6,8,10,12,14])

pop_mean = np.mean(population)
pop_var = np.var(population)

## darwing samples with replacemet 

samples =[]

for sam in itertools.product(population,repeat=3 ):
    samples.append(sam)

sample = np.array(samples)  ## this is our sample 
sample.shape ## this tells us the number of smaples we have generated 

# calculating sample mean and sample varaince of the sample generated 

# first the sample means 
sample_means =[]  ## an empty list to hold all the sample means 

for sample in samples:
    mean = np.mean(sample)
    sample_means.append(mean)

X_bar = np.array(sample_means)


# sample variances 

sample_variances  = []
for  sample  in samples:
    variance = np.var(sample ,ddof = 1)
    sample_variances.append(variance)

S_square = np.array(sample_variances)
```

so now we have the array containng the sample means and the sample variances of all those 125 samples generated namely 
X_bar and S_square . we will now be calculating the E[X_bar]  and var[X_bar]     and similarly for the sample variance S_sqaure 

```python
mean_from_samples = np.mean(X_bar) 
variance_of_xbar = np.var(X_bar)

print(f"the true population mean is {pop_mean}")

print(f"the expected value of the sample mean is {mean_from_samples}")
print(f"the variance of the sample mean is {variance_of_xbar}")

## standard error of xbar 
standard_error_of_xbar = np.sqrt(variance_of_xbar)
print(f"the standard error of sample mean  is {standard_error_of_xbar}")
```

theoraticallly the variance of x bar is sigma square / sample size   if we compute this it turns out to be 8.0 / 3 which is equivalent to 
2.666667 and the expected value of the sample mean is also equal to the population mean which makes the xbar an unbiased estimator

```python
mean_of_ssquare = np.mean(S_square)
var_of_ssquare = np.var(S_square,ddof = 1)

print(f"the true population variance is {pop_var}")

print(f"the expected value of the s_square is {mean_of_ssquare}")
print(f"the variance of s_square is {var_of_ssquare}")

# standard error of sample varaince 
standard_error_of_ssquare = np.sqrt(var_of_ssquare)
print(f"the requried standard error is {standard_error_of_ssquare}")
```
