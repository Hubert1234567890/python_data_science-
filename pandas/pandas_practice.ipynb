```python
import numpy as np
import pandas as pd
country=['india','srilanka','china','usa','brazil'] #square braces
pd.Series(country)
```

```python
scores =[30,89,56,45]
pd.Series(scores)
```

```python
## custom indexing
subject=['maths','english','hindi','science']
marks=[56,78,97,67]
pd.Series(marks,index = subject,name = 'shivam  marks') ## we just use a keyword index
```

```python
## series from dictionary
marks={
    'maths' :98,
    'science':89,
    'english': 93
}
shivam_marks =pd.Series(marks,name='shivam ke marks')
shivam_marks
```

attributes of Series


```python
## size
shivam_marks.size
```

```python
## dtype
shivam_marks.dtype
```

```python
## name attribute (name of a series is its attribute )
shivam_marks.name
```

```python
## is_unique
shivam_marks.is_unique
```

```python
## index
print(shivam_marks.index)
## values
print(shivam_marks.values)
```
