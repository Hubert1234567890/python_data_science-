find the total revenue generated by the company


```python
amount = regs.merge(course, how ='inner',on='course_id')
total = amount.groupby('course_id')['price'].sum()
print(total.sum())
```

calculate the month by month revenue

```python
month_nov = nov.merge(course,how='inner',on='course_id')
nov_rev = month_nov.groupby('course_id')['price'].sum().sum()
print(f"nov_rev is {nov_rev}")
dec_rev = dec.merge(course,how='inner',on='course_id').groupby('course_id')['price'].sum().sum()
print(f"dec_rev is {dec_rev}")
```

```python
months_rev = pd.concat([nov,dec],keys =['Nov','Dec']).reset_index()
nov_dec_rev = months_rev.merge(course,how='inner',on='course_id')
nov_dec_rev.groupby(['level_0'])['price'].sum()

```

print the registration table

plot bar chart for revenue and courses

find the students who have enrolled in both nov and dec

```python
nov_reg = nov['student_id']
dec_reg = dec['student_id']
loyal = np.intersect1d(nov_reg,dec_reg)
student[student['student_id'].isin(loyal)]['name']
```

find the courses that got no enrollment

```python
no_enroll = np.setdiff1d(course['course_id'],regs['course_id'])
course[course['course_id'].isin(no_enroll)]
```

find the students who did not enrolled in any course

```python
no_enroll = np.setdiff1d(student['student_id'],regs['student_id'])
student[student['student_id'].isin(no_enroll)]
```

print student name and their partner name

```python
student.merge(student,how='inner',left_on = 'partner',right_on ='student_id')[['name_x','name_y']]
```

find the top students who got most numbe of enrollments

```python
id = student.merge(regs,how='inner',on = 'student_id').groupby('student_id')['course_id'].count().sort_values(ascending = False).head(3)
student[student['student_id'].isin(id.index)]['name'].sort_values(ascending=True)
```

find the top 3 students who have spent most amount on courses

```python
regs.merge(student,on='student_id').merge(course,on='course_id').groupby(['student_id','name'])['price'].sum().sort_values(ascending=False).head(3)
```
