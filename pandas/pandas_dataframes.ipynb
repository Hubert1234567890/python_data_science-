find the number of movies that start A in each group

```python
genre = imdb.groupby('Genre')
def first_name(groups):
  return groups['Series_Title'].str.startswith('A').sum()


genre.apply(first_name)
```

find ranking of each movies in the group based on the imdb_rating

```python
def ranking_movies(groups):
  groups['rank']= groups['IMDB_Rating'].rank(ascending = False, method = 'dense')
  return groups

genre.apply(ranking_movies)
```

find normlalized imdb rating group wise

```python
def normalized_rating(groups):
  groups['normalized_rating'] = (groups['IMDB_Rating'] - groups['IMDB_Rating'].min())/(groups['IMDB_Rating'].max() - groups['IMDB_Rating'].min())
  return groups

genre.apply(normalized_rating)
```
